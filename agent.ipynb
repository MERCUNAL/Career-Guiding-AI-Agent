```python
from typing_extensions import TypedDict
from typing import Annotated
from langchain_core.messages import AnyMessage
from langgraph.graph.message import add_messages
```

```python
class State(TypedDict):
    messages: Annotated[list[AnyMessage], add_messages]
    user_input: str
    curr_q: str
    completed: bool
    final_result: dict
```

```python
state = {
    "messages": [],
    "user_input": "",
    "curr_q": "What subjects do you enjoy the most?",
    "completed": False,
    "final_result": {}
}
```

```python
from dotenv import load_dotenv
import os
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API")
```

```python
from langchain_groq import ChatGroq
llm  = ChatGroq(model="llama-3.1-8b-instant")
```

```python
prompt = """ You are an AI chatbot assigned to help the user decide their career path. You will ask the user questions only releated to CS Fields.
- The starting question should be "Which field do you like the most in CS?".
- After the user answers, you will ask the next question based on the user's answer.
-You should not ask the user more than 5 questions.
-In the end of the conversation, you will suggest a single career path for the user based on their answers along with the reason by saying "According to the evaluation of your preferences When you are done, start your final answer with:
FINAL ANSWER:"."""
```

```python
def chat(state):
    state["messages"] = [
        {"role": "system", "content": prompt}
    ]
    while not state["completed"]:
        response = llm.invoke(state["messages"])
        print("AI:", response.content)
        state["messages"].append({
            "role": "assistant",
            "content": response.content
        })
        
        user_input = input("User: ")
        state["messages"].append({
            "role": "user",
            "content": user_input
        })
        if "FINAL ANSWER:" in response.content:
            state["completed"] = True
```

```python
state["completed"]=False
chat(state)
```
